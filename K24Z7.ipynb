```python
import pandas as pd
import sklearn

# Importing the file
df = pd.read_csv("Tuned K24Z7(Sheet1).csv")

# Check to see if the file is imported properly
# print("Importing file...")
# print(df.head()) 
# Prints head() properly

# Convert Yes/No to 1/0, and only select rows where Tuner is Genesis
df["Intake"] = df["Intake"].map({"Yes":1,"No":0})
df["Intake Manifold"] = df["Intake Manifold"].map({"Yes":1,"No":0})
df["Downpipe"] = df["Downpipe"].map({"Yes":1,"No":0})
df["Exhaust"] = df["Exhaust"].map({"Yes":1,"No":0})
df = df[df["Tuner"].isin(["Genesis"])]

# Check to see if mapping worked properly
# print("Encoding features...")
# print(df.head())
# Map worked properly

# Set X and Y variables
x = df[["Intake", "Intake Manifold", "Downpipe", "Exhaust"]]
y_HP = df["Horsepower"]
y_TQ = df["Torque"]

# Split the data to training and testing
from sklearn.model_selection import train_test_split
x_train, x_test, y_train_HP, y_test_HP, y_train_TQ, y_test_TQ = train_test_split(
    x, y_HP, y_TQ, test_size=0.2, random_state=42)

# Training the model
from sklearn.ensemble import RandomForestRegressor
# HP Model
model_HP = RandomForestRegressor(n_estimators=500, random_state=42)
model_HP.fit(x_train, y_train_HP)
# TQ Model
model_TQ = RandomForestRegressor(n_estimators=500, random_state=42)
model_TQ.fit(x_train, y_train_TQ)

# Predictions using x_test
from sklearn.metrics import r2_score
predictions_HP = model_HP.predict(x_test)
predictions_TQ = model_TQ.predict(x_test)

# R2 score results for horsepower and torque separately
r2_horsepower = r2_score(y_test_HP, predictions_HP)
r2_torque = r2_score(y_test_TQ, predictions_TQ)
print("Results:")
print("Horsepower R2: ", r2_horsepower)
print("Torque R2: ",r2_torque)

# Seeing which mods have the highest impact for both HP and TQ
importance_HP = pd.Series(model_HP.feature_importances_, index=x.columns)
importance_TQ = pd.Series(model_TQ.feature_importances_, index=x.columns)

# Print Importance for HP
print("\nMod Importance for Horsepower:")
print(importance_HP.sort_values(ascending=False))
# Print Importance for TQ
print("\nMod Importance for Torque:")
print(importance_TQ.sort_values(ascending=False))
```
