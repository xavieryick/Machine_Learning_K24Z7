```python
import pandas as pd
import sklearn

# Importing the file
df = pd.read_csv("Tuned K24Z7(Sheet1).csv")

# Check to see if the file is imported properly
# print("Importing file...")
# print(df.head()) 
# Prints head() properly

# Convert Yes/No to 1/0, and only select rows where Tuner is Genesis
df["Intake"] = df["Intake"].map({"Yes":1,"No":0})
df["Intake Manifold"] = df["Intake Manifold"].map({"Yes":1,"No":0})
df["Downpipe"] = df["Downpipe"].map({"Yes":1,"No":0})
df["Exhaust"] = df["Exhaust"].map({"Yes":1,"No":0})
df = df[df["Tuner"].isin(["Genesis"])]

# Check to see if mapping worked properly
# print("Encoding features...")
# print(df.head())
# Map worked properly

# Set X and Y variables
x = df[["Intake", "Intake Manifold", "Downpipe", "Exhaust"]]
y = df[["Horsepower", "Torque"]]

# Split the data to training and testing
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=42
)

# Training the model
from sklearn.ensemble import RandomForestRegressor
model = RandomForestRegressor(n_estimators=500, random_state=42)
model.fit(x_train, y_train)

# Predictions using x_test
from sklearn.metrics import r2_score
predictions = model.predict(x_test)

# R2 score results for horsepower and torque separately
r2_horsepower = r2_score(y_test["Horsepower"], predictions[:, 0])
r2_torque = r2_score(y_test["Torque"], predictions[:, 1])
print("Results:")
print("Horsepower R2: ", r2_horsepower)
print("Torque R2: ",r2_torque)

# Seeing which mods have the highest impact
importance = pd.Series(model.feature_importances_, index=x.columns)
print("\nMod Importance:")
print(importance.sort_values(ascending=False))
```
